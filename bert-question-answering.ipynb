```python
import torch
from transformers import BertTokenizer, BertForQuestionAnswering


# Load the tokenizer and model
tokenizer = BertTokenizer.from_pretrained(
    "bert-large-uncased-whole-word-masking-finetuned-squad"
)

model = BertForQuestionAnswering.from_pretrained(
    "bert-large-uncased-whole-word-masking-finetuned-squad"
)


# Context used by the question-answering system
context = """
Destiny Rides is an automobile dealership that specializes in selling
new and used vehicles. The company was founded in 2010 and is located
in Lagos, Nigeria. Emhonfubhegbe Destiny is the CEO of Destiny Rides, and the company has a team of experienced salespeople and mechanics who are dedicated to providing excellent customer service. Destiny Rides offers a wide range of vehicles, including sedans, SUVs, trucks, and motorcycles, and also provides financing options for customers. The dealership has built a strong reputation in the local community for its quality vehicles and exceptional service.
"""


def answer_question(question):
    inputs = tokenizer(
        question,
        context,
        return_tensors="pt"
    )

    input_ids = inputs["input_ids"]

    outputs = model(**inputs)

    answer_start = torch.argmax(outputs.start_logits)
    answer_end = torch.argmax(outputs.end_logits)

    if answer_end >= answer_start:
        answer = input_ids[0][answer_start:answer_end + 1]
        answer = tokenizer.decode(
            answer,
            skip_special_tokens=True
        )
        return answer.strip()

    return "I don't know."


# Example
question = "When was Destiny Rides founded?"
answer = answer_question(question)

print("Question:", question)
print("Answer:", answer)
```

```python
question = "who is the CEO of Destiny Rides?"
answer = answer_question(question)
print(answer)
```
